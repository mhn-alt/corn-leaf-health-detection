# Corn Leaf Disease Detection (image + webcam)
Uses the trained `corn_leaf_disease_model.h5`. Put the model file in the same folder as this notebook.

```python
# pip install opencv-python
# pip install tensorflow
# pip install numpy
```

```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
```

```python
# compile=False avoids optimizer/metrics warnings (we only need to predict)
model = load_model('corn_leaf_disease_model.h5', compile=False)

# Training used flow_from_directory -> alphabetical class order: diseased=0, healthy=1
# Sigmoid output = probability of class 1 (healthy)
CLASS_NAMES = ['diseased', 'healthy']
IMG_SIZE = (224, 224)
```

```python
def predict_leaf(frame):
    """frame: BGR image (as read by OpenCV). Returns (label, confidence)."""
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    rgb = cv2.resize(rgb, IMG_SIZE)
    x = preprocess_input(rgb.astype('float32')).reshape(1, 224, 224, 3)
    prob_healthy = float(model.predict(x, verbose=0)[0][0])
    label = CLASS_NAMES[int(prob_healthy > 0.5)]
    confidence = prob_healthy if prob_healthy > 0.5 else 1 - prob_healthy
    return label, confidence


def leaf_detector(frame):
    label, conf = predict_leaf(frame)
    color = (0, 170, 0) if label == 'healthy' else (0, 0, 255)   # BGR
    text = f"{label} ({conf:.0%})"
    h, w = frame.shape[:2]
    cv2.rectangle(frame, (0, 0), (w, 45), color, -1)
    cv2.putText(frame, text, (10, 32), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 255, 255), 2)
    cv2.rectangle(frame, (0, 0), (w - 1, h - 1), color, 4)
    print(text)
    return frame
```

## Test with a jpg image

```python
img = cv2.imread('a2b1b321-40f4-4636-be16-bb69912b6e40___RS_NLB 0798.JPG')   
assert img is not None, "Image not found - check the file name/path"
result = leaf_detector(img)
cv2.imshow("image", result)
cv2.waitKey()
cv2.destroyAllWindows()
```

## Test with webcam (press `q` to quit)

```python
video = cv2.VideoCapture(0)

while True:
    suc, frame = video.read()
    if not suc:
        break

    img = leaf_detector(frame)
    cv2.imshow("WEBCAM", img)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
